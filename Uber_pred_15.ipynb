```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_path = "D:/Uber_pred/yellow_tripdata_2015-01.csv"
```

```python
data = pd.read_csv(data_path)
```

```python
data.sample(5)
```

```python
cols_to_drop = ['extra','payment_type','fare_amount','mta_tax','tip_amount','tolls_amount','improvement_surcharge','total_amount']
```

```python
data_final = data.drop(columns = cols_to_drop)
```

```python
data_final = data_final[
    (data_final['pickup_latitude'].between(40.49, 40.92)) &
    (data_final['pickup_longitude'].between(-74.27, -73.68)) &
    (data_final['dropoff_latitude'].between(40.49, 40.92)) &
    (data_final['dropoff_longitude'].between(-74.27, -73.68))
]

print(f"Rows remaining: {len(data_final)}")
data_final.describe()
```

```python
from sklearn.cluster import MiniBatchKMeans

# Use MiniBatchKMeans (memory-friendly for large data)
kmeans = MiniBatchKMeans(n_clusters=40, random_state=42, batch_size=10000)
data_final['pickup_cluster'] = kmeans.fit_predict(
    data_final[['pickup_latitude', 'pickup_longitude']]
)
data_final['dropoff_cluster'] = kmeans.predict(
    data_final[['dropoff_latitude', 'dropoff_longitude']]
)
```
